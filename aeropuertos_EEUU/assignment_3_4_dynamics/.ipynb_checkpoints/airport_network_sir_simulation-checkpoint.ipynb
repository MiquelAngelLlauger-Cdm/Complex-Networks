
# Airport Network Dynamics — Simple Epidemic Propagation

This notebook adds a **simple dynamical process** on top of the airport network from Assignment 3.

We simulate a basic **SIR epidemic model** on the USA airport network:

- **S** = Susceptible airports
- **I** = Infected airports
- **R** = Recovered airports

Interpretation for airport networks:
- An infected airport represents an airport currently spreading disruption/delay/virus.
- Infection propagates through flight connections.
- Recovery means the airport is no longer spreading.

This is intentionally simple and explainable so it can later be extended into:
- weighted spreading,
- traffic-based transmission,
- cascading failures,
- targeted attacks,
- temporal dynamics,
- multilayer networks.


```

from pathlib import Path
import random

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd

from utils import AirportDatasetConfig, run_pipeline

plt.rcParams['figure.figsize'] = (10, 5)
```


## Load the airport network

This reuses the exact same paths and variable names from the original assignment notebook.


```

# Same configuration as the original notebook
DATA_ROOT = Path('../data')

config = AirportDatasetConfig(
    nodes_path=str(DATA_ROOT / 'reachability-meta.csv' / 'reachability-meta.csv'),
    edges_path=str(DATA_ROOT / 'reachability.txt' / 'reachability.txt'),
    node_id_col='node_id',
    node_name_col='name',
    node_lat_col='latitude',
    node_lon_col='longitude',
    node_pop_col='metro_pop',
    edge_source_col='FromNodeId',
    edge_target_col='ToNodeId',
    edge_weight_col='Weight',
)

result = run_pipeline(config)

G = result['G']
G_undirected = result['G_undirected']

print(nx.info(G_undirected))
```


## Simple SIR simulation

At each timestep:

1. Every infected airport can infect its neighbors with probability `beta`
2. Every infected airport recovers with probability `gamma`

This is one of the most classical spreading models in complex networks.


```

def run_sir_simulation(
    G,
    beta=0.08,
    gamma=0.03,
    initial_infected=None,
    steps=40,
    seed=42,
):
    random.seed(seed)
    np.random.seed(seed)

    nodes = list(G.nodes())

    # Choose starting airport
    if initial_infected is None:
        initial_infected = random.choice(nodes)

    susceptible = set(nodes)
    infected = {initial_infected}
    recovered = set()

    susceptible.remove(initial_infected)

    history = []

    for t in range(steps):

        new_infected = set()
        new_recovered = set()

        # Infection phase
        for node in infected:
            neighbors = list(G.neighbors(node))

            for neighbor in neighbors:
                if neighbor in susceptible:
                    if random.random() < beta:
                        new_infected.add(neighbor)

        # Recovery phase
        for node in infected:
            if random.random() < gamma:
                new_recovered.add(node)

        infected = (infected | new_infected) - new_recovered
        susceptible -= new_infected
        recovered |= new_recovered

        history.append({
            'time': t,
            'susceptible': len(susceptible),
            'infected': len(infected),
            'recovered': len(recovered),
        })

        if len(infected) == 0:
            break

    return pd.DataFrame(history), initial_infected
```


## Start from a highly connected airport

To make the propagation visible, we start from the airport with the highest degree centrality.


```

degree_dict = dict(G_undirected.degree())

initial_airport = max(degree_dict, key=degree_dict.get)

print("Initial infected airport:", initial_airport)
print("Degree:", degree_dict[initial_airport])

sir_df, initial_airport = run_sir_simulation(
    G_undirected,
    beta=0.08,
    gamma=0.03,
    initial_infected=initial_airport,
    steps=50,
    seed=42,
)

sir_df.head()
```


## Epidemic evolution over time


```

plt.plot(sir_df['time'], sir_df['susceptible'], label='Susceptible')
plt.plot(sir_df['time'], sir_df['infected'], label='Infected')
plt.plot(sir_df['time'], sir_df['recovered'], label='Recovered')

plt.xlabel('Time step')
plt.ylabel('Number of airports')
plt.title('SIR dynamics on the airport network')

plt.legend()
plt.grid(True)
plt.show()
```


## Interpretation

Typical behavior:

- Infection initially grows rapidly due to hub connectivity
- The number of infected airports reaches a peak
- Recovery gradually dominates
- Eventually the spreading process dies out

This demonstrates how highly connected transportation networks can rapidly propagate disruptions or diseases.

## Optional extension — compare random vs hub initialization


```

random_start = random.choice(list(G_undirected.nodes()))

sir_random, _ = run_sir_simulation(
    G_undirected,
    beta=0.08,
    gamma=0.03,
    initial_infected=random_start,
    steps=50,
    seed=1,
)

sir_hub, _ = run_sir_simulation(
    G_undirected,
    beta=0.08,
    gamma=0.03,
    initial_infected=initial_airport,
    steps=50,
    seed=1,
)

plt.plot(sir_random['time'], sir_random['infected'], label='Random start')
plt.plot(sir_hub['time'], sir_hub['infected'], label='Hub start')

plt.xlabel('Time step')
plt.ylabel('Infected airports')
plt.title('Effect of initial airport choice')

plt.legend()
plt.grid(True)
plt.show()
```
