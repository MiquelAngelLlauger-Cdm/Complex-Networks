# FAA Air Traffic Control Network (KONECT maayan-faa)

Same analysis workflow as the US reachability study, using `utils_world.py`.

- **Data:** [KONECT maayan-faa](http://konect.cc/networks/maayan-faa) preferred ATC routes (directed, unweighted).
- **Nodes:** integer airport / service-center IDs (no coordinates).
- **Plots:** spring-layout network graphs (no map background).
- **Undirected graph:** each directed edge `u -> v` becomes an undirected link `{u, v}` (e.g. `3->4` connects 3 and 4 even without `4->3`).

```python
from pathlib import Path
import importlib
from IPython.display import display
from bokeh.io import output_notebook, show

import utils_world as utils
importlib.reload(utils)

from utils_world import (
    KONECTGraphConfig,
    run_pipeline_konect,
    plot_all_connections,
    plot_one_way_connections,
    plot_one_way_degree_diff,
    plot_undirected_from_directed,
    plot_centrality_grid,
    plot_degree_histogram,
    plot_local_clustering_map,
    plot_center_periphery,
    plot_degree_distribution_loglog,
    louvain_communities_plot,
    louvain_communities_map_plot,
    plot_louvain_coassignment_heatmaps_pair,
    plot_louvain_community_mean_degrees,
    plot_louvain_community_node_counts,
    plot_louvain_modularity_trajectory,
    louvain_reference_partition,
    louvain_pre_final_communities_map_plot,
)

output_notebook()
```

```python
DATA_ROOT = Path('../data/download.tsv.maayan-faa/maayan-faa')

config = KONECTGraphConfig(
    edges_path=str(DATA_ROOT / 'out.maayan-faa'),
    layout_seed=42,
)

result = run_pipeline_konect(config)
print('KONECT header:', result['konect_header'])

G = result['G']
G_one_way = result['G_one_way']
G_reciprocal_only = result['G_reciprocal_only']
G_undirected = result['G_undirected']

print('Directed summary:')
display(result['directed_table'])
print('Undirected (each directed edge is a link) summary:')
display(result['undirected_table'])
print(
    f"Reciprocal-only subgraph (both directions): {G_reciprocal_only.number_of_nodes()} nodes, "
    f"{G_reciprocal_only.number_of_edges()} edges"
)
```

```python
show(plot_all_connections(G, max_edges=None))
show(plot_one_way_connections(G_one_way, max_edges=None))
show(plot_one_way_degree_diff(G_one_way, max_edges=None))
show(plot_undirected_from_directed(G_undirected, max_edges=None))
```

## Louvain community stability

Repeated Louvain runs, mean degree per community, and modularity trajectory.

```python
N_LOUVAIN_RUNS = 40
LOUVAIN_SEED = 42
HEATMAP_SAMPLE_SEED = 2024
MAX_HEATMAP_NODES = 40

ref_communities, ref_modularity, ref_membership = louvain_reference_partition(
    G_undirected,
    seed=LOUVAIN_SEED,
)

print('Louvain communities (final partition) - node IDs per community:')
for comm_id, community in enumerate(ref_communities, start=1):
    node_ids = sorted(community)
    print(f'\nCommunity {comm_id} ({len(node_ids)} nodes):')
    for node_id in node_ids:
        print(f'  - {node_id}')

map_fig, _, _, _ = louvain_communities_map_plot(
    G_undirected,
    communities=ref_communities,
    modularity_score=ref_modularity,
    seed=LOUVAIN_SEED,
    max_edges=900,
)
show(map_fig)
print(f'Reference Louvain modularity (final): {ref_modularity:.4f}')

co_layout, co_matrices, co_label_groups, co_selected = plot_louvain_coassignment_heatmaps_pair(
    G_undirected,
    communities=ref_communities,
    membership=ref_membership,
    n_runs=N_LOUVAIN_RUNS,
    seed=LOUVAIN_SEED,
    max_nodes=MAX_HEATMAP_NODES,
    heatmap_sample_seed=HEATMAP_SAMPLE_SEED,
)
show(co_layout)
print(
    f'Left: {len(co_label_groups[0])} random high-degree nodes | '
    f'Right: {len(co_label_groups[1])} balanced per community '
    f'({G_undirected.number_of_nodes()} nodes in graph).'
)

deg_fig, community_degrees = plot_louvain_community_mean_degrees(
    G_undirected,
    communities=ref_communities,
)
size_fig, _ = plot_louvain_community_node_counts(
    G_undirected,
    communities=ref_communities,
)
show(deg_fig)
show(size_fig)
display(community_degrees)

mod_fig, modularity_path = plot_louvain_modularity_trajectory(
    G_undirected,
    seed=LOUVAIN_SEED,
)
show(mod_fig)
display(modularity_path)
final_q = modularity_path.loc[
    modularity_path['stage'] == 'final (chosen partition)', 'modularity'
].iloc[0]
pre_final_q = modularity_path.loc[
    modularity_path['stage'].str.contains('pre-final', na=False), 'modularity'
]
print(
    f'Initial Q (singletons): {modularity_path.iloc[0]["modularity"]:.4f} | '
    f'Pre-final Q: {pre_final_q.iloc[0]:.4f} | '
    f'Final Q (chosen): {final_q:.4f}'
)

pre_final_fig, pre_final_modularity, _ = louvain_pre_final_communities_map_plot(
    G_undirected,
    seed=LOUVAIN_SEED,
    max_edges=900,
)
show(pre_final_fig)
print(f'Pre-final level modularity: {pre_final_modularity:.4f}')
```
